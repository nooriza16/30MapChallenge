```python
import osmnx as ox
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

### Hide

```python
def line_bearing(geom):
    (x1, y1), (x2, y2) = geom.coords[0], geom.coords[-1]
    return ox.bearing.calculate_bearing(y1, x1, y2, x2)

gdf["bearing"] = gdf.geometry.apply(line_bearing)

# Add reverse direction
df = pd.DataFrame({"bearing": np.concatenate([gdf["bearing"], (gdf["bearing"] + 180) % 360])})

# Tutorial code
bearings = df['bearing']
ax = bearings.hist(bins=30, zorder=2, alpha=0.3)
xlim = ax.set_xlim(0, 360)
ax.set_title('Network Edge Bearings of Pogung')
plt.show()

n = 30
count, division = np.histogram(bearings, bins=[ang*360/n for ang in range(0, n+1)])
division = division[0:-1]
width = 2 * np.pi / n
ax = plt.subplot(111, projection='polar')
ax.set_theta_zero_location('N')
ax.set_theta_direction('clockwise')
bars = ax.bar(division * np.pi/180 - width * 0.5, count, width=width, bottom=0.0)
ax.set_yticklabels([])          # removes the numbers 50, 100, ... 350
ax.set_title('Network Edge Bearings of Pogung', y=1.1)
plt.show()
```

### Viz

```python

# Build df['bearing'] from the shapefile
gdf = gpd.read_file("Pogung_road.shp").explode(index_parts=False).reset_index(drop=True)
gdf = gdf.to_crs(4326)

#1. Calculate street bearings
def line_bearing(geom):
    (x1, y1), (x2, y2) = geom.coords[0], geom.coords[-1]
    return ox.bearing.calculate_bearing(y1, x1, y2, x2)

gdf["bearing"] = gdf.geometry.apply(line_bearing)

#2. Account for both directions
bearings = pd.concat([
    gdf["bearing"],
    (gdf["bearing"] + 180) % 360
])

#3. Plot the histogram
ax = bearings.hist(
    bins=30,
    zorder=2,
    alpha=0.8,
    edgecolor="none"
)

ax.set_xlim(0, 360)
ax.set_title("Street Network Orientation of Pogung")
ax.set_xlabel("Street Bearing (degrees)")
ax.set_ylabel("n of street segments")

# Remove gridlines
ax.grid(False)

plt.show()

```

```python
plt.savefig(
    "street_orientation.png",
    dpi=300,
    bbox_inches="tight",
    transparent=False
)
```

```python
#Plot the polar histogram
n = 30
count, division = np.histogram(
    bearings,
    bins=np.linspace(0, 360, n + 1)
)
width = 2 * np.pi / n

ax = plt.subplot(111, projection="polar")
ax.set_theta_zero_location("N")
ax.set_theta_direction("clockwise")
ax.bar(
    np.deg2rad(division[:-1]),
    count,
    width=width
)
ax.set_yticklabels([])
ax.set_title("Street Network Orientation of Pogung", y=1.1)

plt.savefig(
    "street_polar.png",
    dpi=300,
    bbox_inches="tight",
    transparent=False
)
plt.show()
```

```python
import osmnx as ox
import matplotlib.pyplot as plt

# 1. Download street network
G = ox.graph_from_place(
    "Pogung, Yogyakarta, Indonesia",
    network_type="drive"
)

# 2. Convert to undirected graph
Gu = ox.convert.to_undirected(G)

# 3. Calculate edge bearings
Gu = ox.bearing.add_edge_bearings(Gu)

# 4. Plot street orientation
fig, ax = ox.plot.plot_orientation(
    Gu,
    title="Street Network Orientation of Pogung",
    area=True
)


plt.show()
```
