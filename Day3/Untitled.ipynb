```python
# Use a mirror and fail fast instead of waiting 180 s
ox.settings.overpass_url = "https://overpass.kumi.systems/api"   # osmnx 2.x
# ox.settings.overpass_endpoint = "https://overpass.kumi.systems/api"  # osmnx 1.x
ox.settings.requests_timeout = 60
ox.settings.use_cache = True
ox.settings.log_console = True   # shows what it's doing

point = (51.52462627696973, -0.1344432605972362)
dist = 600   # start small, then increase once it works

buildings = ox.features.features_from_point(
    point, tags={"building": True}, dist=dist
)

# Keep only what you need and save locally so you never re-download
buildings = buildings[["geometry", "building", "height", "building:levels", "name"]
                      .__class__(c for c in ["geometry", "building", "height", "building:levels", "name"]
                                 if c in buildings.columns)]
buildings = buildings[buildings.geom_type.isin(["Polygon", "MultiPolygon"])]
buildings.to_file("buildings_600m.gpkg", driver="GPKG")
```
